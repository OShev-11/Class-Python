```python
import requests
```

```python
class Rate:
    def __init__(self, format_='name, value'):
      self.format = format_

    def exchange_rates(self):
        self.r = requests.get('https://www.cbr-xml-daily.ru/daily_json.js')
        return self.r.json()['Valute']

    def make_format(self, currency):
        response = self.exchange_rates()

        if currency in response:
            if self.format == 'full':
                return response[currency]
            if self.format == 'name, value':
                name = response[currency]['Name']
                value = response[currency]['Value']
                return f"{name}: {value}"

        return 'Error'

r = Rate(format_='name, value')
r.make_format('EUR')
```

```python
class Rate:
    def __init__(self, format_='value_diff'):
      self.format = format_

    def exchange_rates(self):
        self.r = requests.get('https://www.cbr-xml-daily.ru/daily_json.js')
        return self.r.json()['Valute']

    def make_format(self, currency):
        response = self.exchange_rates()
        name = name = response[currency]['Name']
        diff = response[currency]['Value'] - response[currency]['Previous']

        if currency in response:
            if self.format == 'full':
                return response[currency]
            if self.format == 'value_diff':
                return f"Изменение {name}: {diff}"
            if self.format == 'value':
                return response[currency]['Value']

        return 'Error'


r = Rate(format_='value_diff')
r.make_format('EUR')
```
